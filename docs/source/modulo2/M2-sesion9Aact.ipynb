````python
# Importamos pgmpy.factors.discrete.DiscreteFactor
from pgmpy.factors.discrete import DiscreteFactor
````

````python
phi1=DiscreteFactor(['X', 'Y'], [2, 2], [0.8, 0.5, 0.5, 0.6])
phi2=DiscreteFactor(['Y', 'Z'], [2, 2], [0.2, 0.2, 0.9, 1])
phi3=phi1*phi2
print(phi3)
````

````python
phi_reduced = phi3.reduce(values=[('X',0),('Y',0),('Z',0)], inplace=False)
print(phi_reduced)
````

````python
phi_reduced2 = phi3.reduce(values=[('X',0),('Y',1),('Z',1)], inplace=False)
print(phi_reduced2)
````

### Actividad: parámetros de una red bayesiana y la conjunta

Considera esta pequeña **red bayesiana** de 4 variables:

| Variable | Valores | Padres |
| -------- | ------- | ------ |
| $C$ (Clima) | soleado, nublado, lluvioso &nbsp;(3) | — |
| $T$ (Tráfico) | bajo, alto &nbsp;(2) | $C$ |
| $A$ (Accidente) | sí, no &nbsp;(2) | $T$ |
| $L$ (Llego tarde) | sí, no &nbsp;(2) | $T$ |

Estructura del grafo:

```
   C
   |
   v
   T
  / \
 v   v
 A   L
```

**Responde:**

1. Escribe la **factorización** de la conjunta $P(C, T, A, L)$ usando la **regla de la cadena** + las **independencias locales** (cada variable dada *solo sus padres*).

2. ¿Cuántos **parámetros independientes** necesita **cada** factor (CPD)? ¿Y la **red completa** (la suma)?
   > *Recuerda:* $\;\#\text{params}\big(P(X\mid \mathrm{Pa})\big) = \big(|\mathrm{Val}(X)| - 1\big)\times \prod_{P\in \mathrm{Pa}}|\mathrm{Val}(P)|$

3. ¿Cuántos parámetros necesitaría la **distribución conjunta directa** (sin factorizar)?

4. **Compara:** ¿cuántos parámetros te ahorras al factorizar?



**Respuestas**
1. P(C,T,A,L)=P(C)P(T|C)P(A|T)P(L|T)
2. P(C)=3-1=2, P(T|C)=(2-1)*3=3, P(A|T)=(2-1)*2=2, P(L|T)=(2-1)*2=2

| Factor | Calculo | Parms |
| ------ | ------- | ----- |
| P(C) | 3-1 | 2 |
| P(T/C) | (2-1)*3 | 3 |
| P(A/T) | (2-1)*2 | 2 |
| P(L/T) | (2-1)*2 | 2 |

La red completa tiene 2+3+2+2=9 parametros

3. La distribucion conjunta sin factorizar necesitaria (3x2x2x2)=24-1=23 parametros
4. Se ahorran 23-9=14 parametros
